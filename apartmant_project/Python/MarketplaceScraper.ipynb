```python
from bs4 import BeautifulSoup
import pandas as pd
import requests
import time

def extract(page = ""):
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/146.0.0.0 Safari/537.36"}
    url = f"https://www.property24.com/to-rent/pretoria/gauteng/1/{page}"
    
    url_page = requests.get(url, headers=headers)
    get_html = BeautifulSoup(url_page.text, "html.parser")
    
    return get_html
```

```python
html = extract()
```
